```python
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.datasets import mnist, fashion_mnist
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Conv2D, Dense, Flatten, MaxPooling2D
from tensorflow.keras.callbacks import Callback

tf.keras.utils.set_random_seed(42)   # make results as repeatable as possible
print('TensorFlow', tf.__version__, '| GPU:', tf.config.list_physical_devices('GPU') or 'none (CPU)')

EPOCHS, BATCH_SIZE = 5, 64           # same settings as the book
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

# Helper functions
def prepare_data(X_train, y_train, X_test, y_test):
    """Reshape to (28, 28, 1), normalize to 0.0-1.0, one-hot encode labels (16.6.3)."""
    X_train = X_train.reshape((60000, 28, 28, 1)).astype('float32') / 255
    X_test = X_test.reshape((10000, 28, 28, 1)).astype('float32') / 255
    return X_train, to_categorical(y_train), X_test, to_categorical(y_test)


def build_cnn(variant='baseline'):
    """The chapter's convnet (16.6.4).
    'baseline'      : Conv(64)/Pool, Conv(128)/Pool, Flatten, Dense(128), Dense(10)
    'no_dense128'   : 16.4(a) first Dense(128) layer removed
    'add_dense4096' : 16.4(b) Dense(4096) added before the two Dense layers
    """
    cnn = Sequential()
    cnn.add(Input(shape=(28, 28, 1)))
    cnn.add(Conv2D(filters=64, kernel_size=(3, 3), activation='relu'))
    cnn.add(MaxPooling2D(pool_size=(2, 2)))
    cnn.add(Conv2D(filters=128, kernel_size=(3, 3), activation='relu'))
    cnn.add(MaxPooling2D(pool_size=(2, 2)))
    cnn.add(Flatten())
    if variant == 'add_dense4096':
        cnn.add(Dense(units=4096, activation='relu'))
    if variant != 'no_dense128':
        cnn.add(Dense(units=128, activation='relu'))
    cnn.add(Dense(units=10, activation='softmax'))
    cnn.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    return cnn


class EpochTimer(Callback):
    """Records how long each epoch takes."""
    def on_train_begin(self, logs=None): self.times = []
    def on_epoch_begin(self, epoch, logs=None): self.t = time.perf_counter()
    def on_epoch_end(self, epoch, logs=None): self.times.append(time.perf_counter() - self.t)


def train_and_evaluate(cnn, X_train, y_train, X_test, y_test):
    timer = EpochTimer()
    start = time.perf_counter()
    history = cnn.fit(X_train, y_train, epochs=EPOCHS, batch_size=BATCH_SIZE,
                      validation_split=0.1, callbacks=[timer], verbose=2)
    train_time = time.perf_counter() - start
    loss, accuracy = cnn.evaluate(X_test, y_test, verbose=0)
    cnn.predict(X_test[:64], verbose=0)                    # warm-up
    start = time.perf_counter()
    predictions = cnn.predict(X_test, verbose=0)
    predict_time = time.perf_counter() - start
    return {'acc': accuracy, 'loss': loss, 'train_time': train_time,
            # median epoch time ignores the slow first epoch / random hiccups
            'epoch_time': float(np.median(timer.times)),
            'predict_time': predict_time, 'params': cnn.count_params(),
            'history': history.history, 'predictions': predictions}

# Fashion-MNIST

(X_train_f, y_train_f), (X_test_f, y_test_f) = fashion_mnist.load_data()
print('\nFashion-MNIST shapes:', X_train_f.shape, y_train_f.shape, X_test_f.shape, y_test_f.shape)

# Show 24 sample images
index = np.random.choice(np.arange(len(X_train_f)), 24, replace=False)
fig, axes = plt.subplots(nrows=4, ncols=6, figsize=(16, 9))
for ax, image, target in zip(axes.ravel(), X_train_f[index], y_train_f[index]):
    ax.imshow(image, cmap=plt.cm.gray_r); ax.set_xticks([]); ax.set_yticks([])
    ax.set_title(class_names[target])
plt.suptitle('Fashion-MNIST training samples'); plt.tight_layout(); plt.show()

fashion_labels = y_test_f.copy()
X_train_f, y_train_f, X_test_f, y_test_f = prepare_data(X_train_f, y_train_f, X_test_f, y_test_f)

# Load and prepare MNIST for the comparison
(X_train_m, y_train_m), (X_test_m, y_test_m) = mnist.load_data()
mnist_labels = y_test_m.copy()
X_train_m, y_train_m, X_test_m, y_test_m = prepare_data(X_train_m, y_train_m, X_test_m, y_test_m)

data = {'MNIST': (X_train_m, y_train_m, X_test_m, y_test_m),
        'Fashion-MNIST': (X_train_f, y_train_f, X_test_f, y_test_f)}

build_cnn('baseline').summary()

results = {}
for ds in ['Fashion-MNIST', 'MNIST']:
    print(f'\n===== {ds} / book baseline model =====')
    tf.keras.utils.set_random_seed(42)
    results[(ds, 'baseline')] = train_and_evaluate(build_cnn('baseline'), *data[ds])
    r = results[(ds, 'baseline')]
    print(f"{ds}: test accuracy {r['acc']:.2%}, total training {r['train_time']:.1f} s, "
          f"~{r['epoch_time']:.1f} s/epoch")

# first question, Which Fashion-MNIST items does the model get wrong?
pred_f = np.argmax(results[('Fashion-MNIST', 'baseline')]['predictions'], axis=1)
pred_m = np.argmax(results[('MNIST', 'baseline')]['predictions'], axis=1)
wrong_f = np.where(pred_f != fashion_labels)[0]
wrong_m = np.where(pred_m != mnist_labels)[0]
print(f'\nIncorrect predictions: MNIST {len(wrong_m)}, Fashion-MNIST {len(wrong_f)} (out of 10,000)')

images = X_test_f.reshape((10000, 28, 28))
fig, axes = plt.subplots(nrows=4, ncols=6, figsize=(16, 12))
for ax, i in zip(axes.ravel(), wrong_f[:24]):
    ax.imshow(images[i], cmap=plt.cm.gray_r); ax.set_xticks([]); ax.set_yticks([])
    ax.set_title(f'index: {i}\np: {class_names[pred_f[i]]}\ne: {class_names[fashion_labels[i]]}')
plt.suptitle('Fashion-MNIST incorrect predictions (p = predicted, e = expected)'); plt.tight_layout(); plt.show()

per_class = pd.Series([(pred_f[fashion_labels == c] == c).mean() for c in range(10)],
                      index=class_names).sort_values()
print('\nFashion-MNIST accuracy per class:'); print(per_class.map('{:.1%}'.format).to_string())

confusions = pd.Series([f'{class_names[e]} -> {class_names[p]}'
                        for e, p in zip(fashion_labels[wrong_f], pred_f[wrong_f])]).value_counts()
print('\nMost common mistakes (expected -> predicted):'); print(confusions.head(6).to_string())

#Comparison table and training curves
table_161 = pd.DataFrame({ds: {'test accuracy': f"{results[(ds,'baseline')]['acc']:.2%}",
                               'errors / 10,000': len(wrong_m) if ds == 'MNIST' else len(wrong_f),
                               'final train acc': f"{results[(ds,'baseline')]['history']['accuracy'][-1]:.2%}",
                               'final val acc': f"{results[(ds,'baseline')]['history']['val_accuracy'][-1]:.2%}",
                               'total train time (s)': f"{results[(ds,'baseline')]['train_time']:.1f}",
                               'median s / epoch': f"{results[(ds,'baseline')]['epoch_time']:.1f}"}
                          for ds in ['MNIST', 'Fashion-MNIST']})
print('\n16.1 COMPARISON'); print(table_161.to_string())

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
for ds, color in [('MNIST', 'tab:blue'), ('Fashion-MNIST', 'tab:orange')]:
    h = results[(ds, 'baseline')]['history']; ep = range(1, EPOCHS + 1)
    axes[0].plot(ep, h['accuracy'], '--', color=color, label=f'{ds} train')
    axes[0].plot(ep, h['val_accuracy'], '-o', color=color, label=f'{ds} validation')
    axes[1].plot(ep, h['loss'], '--', color=color, label=f'{ds} train')
    axes[1].plot(ep, h['val_loss'], '-o', color=color, label=f'{ds} validation')
axes[0].set_title('Accuracy per epoch'); axes[1].set_title('Loss per epoch')
for ax in axes: ax.set_xlabel('epoch'); ax.legend()
plt.tight_layout(); plt.show()

# Remove / add Dense layers
print('\n16.4(a) model:'); build_cnn('no_dense128').summary()
print('\n16.4(b) model:'); build_cnn('add_dense4096').summary()

for ds in ['MNIST', 'Fashion-MNIST']:
    for variant in ['no_dense128', 'add_dense4096']:
        print(f'\n===== {ds} / {variant} =====')
        tf.keras.utils.set_random_seed(42)
        results[(ds, variant)] = train_and_evaluate(build_cnn(variant), *data[ds])
        r = results[(ds, variant)]
        print(f"test accuracy {r['acc']:.2%}, ~{r['epoch_time']:.1f} s/epoch, predict 10k: {r['predict_time']:.2f} s")

names = {'no_dense128': '(a) Dense(128) removed', 'baseline': 'Book baseline', 'add_dense4096': '(b) + Dense(4096)'}
order = ['no_dense128', 'baseline', 'add_dense4096']
rows = []
for ds in ['MNIST', 'Fashion-MNIST']:
    b = results[(ds, 'baseline')]
    for v in order:
        r = results[(ds, v)]
        rows.append({'dataset': ds, 'model': names[v], 'parameters': f"{r['params']:,}",
                     'test acc': f"{r['acc']:.2%}",
                     'Δ acc (pts)': f"{(r['acc'] - b['acc']) * 100:+.2f}",
                     's / epoch': f"{r['epoch_time']:.1f}",
                     'train × base': f"{r['epoch_time'] / b['epoch_time']:.2f}×",
                     'predict 10k (s)': f"{r['predict_time']:.2f}",
                     'predict × base': f"{r['predict_time'] / b['predict_time']:.2f}×"})
print('\n16.4 COMPARISON'); print(pd.DataFrame(rows).set_index(['dataset', 'model']).to_string())

fig, axes = plt.subplots(1, 3, figsize=(17, 4.5))
x = np.arange(3); w = 0.38
for k, (metric, title) in enumerate([('acc', 'Test accuracy'), ('epoch_time', 'Seconds per epoch'),
                                     ('predict_time', 'Predict 10,000 images (s)')]):
    for j, (ds, color) in enumerate([('MNIST', 'tab:blue'), ('Fashion-MNIST', 'tab:orange')]):
        bars = axes[k].bar(x + (j - 0.5) * w, [results[(ds, v)][metric] for v in order], w, label=ds, color=color)
        axes[k].bar_label(bars, fmt='%.4f' if metric == 'acc' else '%.1f', fontsize=8)
    axes[k].set_xticks(x, [names[v] for v in order], fontsize=9); axes[k].set_title(title)
axes[0].set_ylim(0.85, 1.0); axes[0].legend(); plt.tight_layout(); plt.show()

# answers (generated from the numbers in THIS run)
def pts(a, b): return (a - b) * 100
def change(d): return 'about the same' if abs(d) < 0.3 else ('higher' if d > 0 else 'lower')

m, f = results[('MNIST', 'baseline')], results[('Fashion-MNIST', 'baseline')]
gap_f = f['history']['accuracy'][-1] - f['history']['val_accuracy'][-1]
gap_m = m['history']['accuracy'][-1] - m['history']['val_accuracy'][-1]
ratio = f['epoch_time'] / m['epoch_time']
top = confusions.index[:3].tolist()

print('\n' + '=' * 70 + '\n  exercise 16.1\n' + '=' * 70)
print(f"""
How well does the model perform on Fashion-MNIST compared to MNIST?
  The same convnet reaches {f['acc']:.2%} test accuracy on Fashion-MNIST versus
  {m['acc']:.2%} on MNIST, {pts(m['acc'], f['acc']):.1f} percentage points lower, and it makes
  {len(wrong_f)} mistakes instead of {len(wrong_m)} (~{len(wrong_f)/max(len(wrong_m),1):.0f}x as many).
  Fashion-MNIST is a harder problem even though the images have the same format. The weakest class is
  '{per_class.index[0]}' ({per_class.iloc[0]:.0%} correct), and the most common mistakes
  are {', '.join(top)}. Items with a distinctive shape (e.g. '{per_class.index[-1]}',
  {per_class.iloc[-1]:.0%}) are easy. Train vs. validation accuracy gap after the last
  epoch: Fashion-MNIST {gap_f*100:.1f} pts, MNIST {gap_m*100:.1f} pts.""" +
("\n  The gap is larger on Fashion-MNIST, so the model is starting to overfit there."
 if gap_f > gap_m and gap_f > 0 else "") + "\n" + f"""
How do the training times compare?
  ~{m['epoch_time']:.1f} s/epoch for MNIST vs ~{f['epoch_time']:.1f} s/epoch for Fashion-MNIST
  (ratio {ratio:.2f}), i.e. essentially the same.
  Both datasets have exactly 60,000 training images of 28x28x1 pixels, and the
  network is identical, so each epoch does the same amount of computation. Fashion-MNIST is harder to learn,
  not slower to train.
""")

print('=' * 70 + '\nANSWER - EXERCISE 16.4\n' + '=' * 70)
for ds in ['MNIST', 'Fashion-MNIST']:
    b, a, c = results[(ds, 'baseline')], results[(ds, 'no_dense128')], results[(ds, 'add_dense4096')]
    print(f"""{ds}:
   a) Dense(128) removed: accuracy {a['acc']:.2%} vs {b['acc']:.2%} ({pts(a['acc'], b['acc']):+.2f} pts, {change(pts(a['acc'], b['acc']))}),
      {a['epoch_time']/b['epoch_time']:.2f}x training time, {a['predict_time']/b['predict_time']:.2f}x prediction time.
  b) Dense(4096) added:  accuracy {c['acc']:.2%} vs {b['acc']:.2%} ({pts(c['acc'], b['acc']):+.2f} pts, {change(pts(c['acc'], b['acc']))}),
      {c['epoch_time']/b['epoch_time']:.2f}x training time, {c['predict_time']/b['predict_time']:.2f}x prediction time,
      {c['params']/b['params']:.0f}x as many parameters ({b['params']:,} -> {c['params']:,}).
""")
d_a = np.mean([pts(results[(d, 'no_dense128')]['acc'], results[(d, 'baseline')]['acc']) for d in data])
d_b = np.mean([pts(results[(d, 'add_dense4096')]['acc'], results[(d, 'baseline')]['acc']) for d in data])
slow_b = np.mean([results[(d, 'add_dense4096')]['epoch_time'] / results[(d, 'baseline')]['epoch_time'] for d in data])

print("Explanation:")
print(f"""  a) Removing the first Dense layer leaves only a linear softmax layer on top of
      the 3,200 convolutional features, so the model can no longer learn non-linear
      combinations of them. The two convolution layers still do most of the work, so
      the effect on accuracy is small (average {d_a:+.2f} pts in this run{
      ', a slight drop' if d_a < -0.1 else ', within normal run-to-run variation'}).
      The model is ~78% smaller (485,514 -> 106,506 parameters) and slightly faster.""")
print(f"""  b) A 4096-neuron layer adds ~13 million weights (3,200 x 4,096). That size suits
      pretrained ImageNet models such as VGG16 (224x224 color images, 1,000 classes),
      but it is much more capacity than 28x28 grayscale images with 10 classes need.
      Accuracy changed by an average of {d_b:+.2f} pts{
      ' (essentially no change)' if abs(d_b) < 0.5 else ' (a small change)'}, while training
      was ~{slow_b:.1f}x slower and prediction was also slower. The accuracy gain, if any,
      is not worth the large cost in speed and model size, so for these datasets the
      book's single Dense(128) layer is the better trade-off.
""")
```
